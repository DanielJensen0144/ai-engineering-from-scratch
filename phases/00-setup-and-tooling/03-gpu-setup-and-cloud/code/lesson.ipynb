# Build it: GPU vs CPU Benchmark

```python
import torch
import time

n_dims = 2**13

a_cpu = torch.randn(n_dims, n_dims)
b_cpu = torch.randn(n_dims, n_dims)

start = time.time()
c_cpu = a_cpu @ b_cpu
cpu_time = time.time() - start

print(f"CPU: {cpu_time:.3f}s")

if torch.backends.mps.is_available():
    a_gpu = a_cpu.to('mps')
    b_gpu = b_cpu.to('mps')

    torch.mps.synchronize()
    start = time.time()
    c_gpu = a_gpu @ b_gpu
    torch.mps.synchronize()
    gpu_time = time.time() - start

    print(f"GPU: {gpu_time:.3f}s")
    print(f"Speedup: {cpu_time/gpu_time:.2f}x")
```

### Remark
I tested this in google colab using a cpu and a T4 gpu, the cpu took like 12-16 seconds whereas the gpu took 0.404 seconds, the conclusion from this is that my gpu isn't slow, it is normal, rather, the surprise is that my cpu is really fast.
